```python
# Step 3: Data Cleaning and Preprocessing
# Drop rows with missing essential values
apps_df = apps_df.dropna(subset=['Installs', 'Price', 'Size', 'Android Ver', 'Content Rating', 'App', 'Type'])
```

```python
# Clean 'Installs' column
apps_df['Installs'] = apps_df['Installs'].str.replace(r'[+,]', '', regex=True).astype(float)
```

```python
# Clean 'Price' column
apps_df['Price'] = apps_df['Price'].str.replace(r'[$]', '', regex=True).astype(float)
```

```python
# Convert 'Size' to KB
def convert_size_to_kb(size):
    if isinstance(size, str):
        size = size.lower()
        if 'm' in size:
            size = size.replace('m', '')
            try:
                return float(size) * 1024
            except ValueError:
                return None
        elif 'k' in size:
            size = size.replace('k', '')
            try:
                return float(size)
            except ValueError:
                return None
        else:
            return None
    return size 
```

```python
# Handle numeric sizes directly
apps_df['Size'] = apps_df['Size'].replace('Varies with device', None)
apps_df['Size'] = apps_df['Size'].apply(convert_size_to_kb).astype(float)
```

```python
# Extract numeric Android version
apps_df['Android Ver'] = apps_df['Android Ver'].str.extract(r'(\d+\.\d+)', expand=False).astype(float)
```

```python
#Step 4: Apply Filters
filtered_apps = apps_df[
    (apps_df['Installs'] >= 10000) &
    (apps_df['Size'] > 15 * 1024) &  # Size > 15MB in KB
    (apps_df['Android Ver'] > 4.0) &
    (apps_df['Content Rating'] == 'Everyone') &
    (apps_df['App'].str.len() <= 30)
].copy()
```

```python
# Step 5: Calculate Revenue for Paid Apps
filtered_apps['Revenue'] = filtered_apps.apply(
    lambda row: row['Price'] * row['Installs'] if row['Type'] == 'Paid' else 0, axis=1
)
```

```python
# Step 6: Identify Top 3 Categories
top_categories = filtered_apps['Category'].value_counts().nlargest(3).index
top_category_apps = filtered_apps[filtered_apps['Category'].isin(top_categories)]
```

```python
# Step 7: Group Data by Category and Type
grouped_data = top_category_apps.groupby(['Category', 'Type']).agg(
    avg_installs=('Installs', 'mean'),
    avg_revenue=('Revenue', 'mean')
).reset_index()
```

```python
# Step 8: Time-Based Restriction for Plotting
ist = pytz.timezone('Asia/Kolkata')
current_time = datetime.now(ist)

if current_time.hour == 13:  # 1 PM in 24-hour format
    # Plotting
    fig, ax1 = plt.subplots(figsize=(10, 6))

    # Bar plot for average installs
    sns.barplot(x='Category', y='avg_installs', hue='Type', data=grouped_data, ax=ax1, palette='Blues')
    ax1.set_ylabel('Average Installs', color='blue')
    ax1.set_xlabel('App Category')
    ax1.tick_params(axis='y', labelcolor='blue')

    # Line plot for average revenue
    ax2 = ax1.twinx()
    sns.lineplot(
        x='Category', y='avg_revenue', hue='Type', data=grouped_data, markers='o', dashes=False, palette='Greens', ax=ax2
    )
    ax2.set_ylabel('Average Revenue ($)', color='green')
    ax2.tick_params(axis='y', labelcolor='green')

    # Add legends and title
    ax1.legend(loc='upper left', title='Type (Installs)')
    ax2.legend(loc='upper right', title='Type (Revenue)')

    plt.title('Comparison of Average Installs and Revenue for Free vs. Paid Apps in Top 3 Categories')
    plt.tight_layout()
    plt.show()
else:
    print("The graph is only available between 1 PM and 2 PM IST.")
```
